The car sensors take a reading abt every 0.1-0.25m of track, depending on the speed. if the grid points are 1m apart, they skip readings in between, so details like the exact spot where braking starts get blurred.

s itself is only accurate to within roughly a meter, putting grid points 0.05m apart doesnt make the data more accurate. we will just interpolate make up values between real readings.

a sweet spot would be 0.3m, it's close to the rate of the readings and it doesnt claim more precision than 2 really has. so the details are kept without inventing any.

the lap where the driver boxed will be dropped since it's mostly standstill, the rows where the car has stopped (the ones we suggested to be blue flags) will be dropped as well since they do not contribute to our analysis goals.

```python
import numpy as np
import plotly.graph_objects as go
from src.loader import load_centerline

K_STRAIGHT = 0.021     # |curvature| below this = straight-ish
MIN_STRAIGHT_M = 20    # straight-ish stretch must be at least this long
K_CORNER = 0.0385       # |curvature| above this = clearly a corner
SMOOTH_WINDOW_M = 5   # rolling-mean window (odd number, 1 point = 1 m)

def smooth_circular(values: np.ndarray, window: int):
    """Rolling mean over `window` points, wrapping around the end of the loop."""
    half = window // 2
    padded = np.concatenate([values[-half:], values, values[:half]])
    return np.convolve(padded, np.ones(window) / window, mode="valid")

cl = load_centerline()
cl["curvature_smooth"] = smooth_circular(cl["curvature"].to_numpy(), SMOOTH_WINDOW_M)

# straights: low smoothed curvature for at least MIN_STRAIGHT_M
low = (cl["curvature_smooth"].abs() < K_STRAIGHT).to_numpy()
run_id = np.concatenate([[0], np.cumsum(low[1:] != low[:-1])])
if low[0] == low[-1] and run_id[-1] != 0:          # loop: join first and last run
    run_id[run_id == run_id[-1]] = 0
run_len = np.bincount(run_id)
cl["is_straight"] = low & (run_len[run_id] >= MIN_STRAIGHT_M)

fig = go.Figure()
for is_straight, colour, name in [(True, "blue", "straight"), (False, "red", "corner")]:
    part = cl[cl["is_straight"] == is_straight]
    fig.add_trace(go.Scatter(
        x=part["x"], y=part["y"],
        mode="markers",
        marker=dict(size=4, color=colour),
        customdata=part[["s", "curvature_smooth"]],
        hovertemplate="s=%{customdata[0]:.0f} m<br>curvature=%{customdata[1]:.04f}<extra></extra>",
        name=name,
    ))

# sharp points on top
sharp = cl[cl["curvature_smooth"].abs() > K_CORNER]
fig.add_trace(go.Scatter(
    x=sharp["x"], y=sharp["y"],
    mode="markers",
    marker=dict(size=3, color="black", symbol="x"),
    hoverinfo="skip",
    showlegend=False,
))

fig.update_yaxes(scaleanchor="x", scaleratio=1)
fig.update_layout(title=f"Straights vs corners (smoothed {SMOOTH_WINDOW_M} m, "
                        f"|curvature| < {K_STRAIGHT} for ≥ {MIN_STRAIGHT_M} m)",
                  xaxis_title="x (m, east)", yaxis_title="y (m, north)",
                  height=700, width=1000,
                  legend=dict(orientation="h", y=-0.1))
fig.show()
```

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from src.loader import load_centerline

K_STRAIGHT = 0.021     # |curvature| below this = straight-ish
MIN_STRAIGHT_M = 20    # straight-ish stretch must be at least this long
K_CORNER = 0.0385      # |curvature| above this = clearly a corner
SMOOTH_WINDOW_M = 5    # rolling-mean window (odd number, 1 point = 1 m)



def runs_of(mask: np.ndarray) :
    """Number each run of consecutive equal values; first and last run joined (track is a loop)."""
    run_id = np.concatenate([[0], np.cumsum(mask[1:] != mask[:-1])])
    if mask[0] == mask[-1] and run_id[-1] != 0:
        run_id[run_id == run_id[-1]] = 0
    return run_id

cl = load_centerline()
cl["curvature_smooth"] = smooth_circular(cl["curvature"].to_numpy(), SMOOTH_WINDOW_M)
curv = cl["curvature_smooth"].abs().to_numpy()      # |curvature|, one value per metre

# straights: low curvature for at least MIN_STRAIGHT_M
low = curv < K_STRAIGHT
low_runs = runs_of(low)
cl["is_straight"] = low & (np.bincount(low_runs)[low_runs] >= MIN_STRAIGHT_M)

# apexes: one per run of high curvature, at its sharpest point
high = curv > K_CORNER
cl["high_run"] = runs_of(high)
apex_idx = cl[high].groupby("high_run")["curvature_smooth"].apply(lambda c: c.abs().idxmax())
apexes = cl.loc[apex_idx].sort_values("s").reset_index(drop=True)
apexes["corner"] = np.arange(1, len(apexes) + 1)
apexes["direction"] = np.where(apexes["curvature_smooth"] > 0, "L", "R")

# plot
fig = go.Figure()
for is_straight, colour, name in [(True, "blue", "straight"), (False, "red", "corner")]:
    part = cl[cl["is_straight"] == is_straight]
    fig.add_trace(go.Scatter(
        x=part["x"], y=part["y"], mode="markers",
        marker=dict(size=4, color=colour),
        customdata=part[["s", "curvature_smooth"]],
        hovertemplate="s=%{customdata[0]:.0f} m<br>curvature=%{customdata[1]:.04f}<extra></extra>",
        name=name,
    ))

sharp = cl[high]
fig.add_trace(go.Scatter(
    x=sharp["x"], y=sharp["y"], mode="markers",
    marker=dict(size=3, color="black", symbol="x"),
    hoverinfo="skip", showlegend=False,
))

fig.add_trace(go.Scatter(
    x=apexes["x"], y=apexes["y"], mode="markers+text",
    marker=dict(size=9, color="gold", line=dict(width=1, color="black")),
    text=apexes["corner"].astype(str) + apexes["direction"],
    textposition="top center",
    customdata=apexes[["s", "curvature_smooth"]],
    hovertemplate="apex s=%{customdata[0]:.0f} m<br>curvature=%{customdata[1]:.04f}<extra></extra>",
    name="apex",
))

fig.update_yaxes(scaleanchor="x", scaleratio=1)
fig.update_layout(title=f"Corners (smoothed {SMOOTH_WINDOW_M} m, peak > {K_CORNER})",
                  xaxis_title="x (m, east)", yaxis_title="y (m, north)",
                  height=700, width=1000, legend=dict(orientation="h", y=-0.1))
fig.show()
```

looking at the graph and the table, i think two apexes belng to the same corner if these conditions apply:
1. they have the same direction
2. the min curvature between them is >= 0.03 
3. the distance between them is <= 35m

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from src.loader import load_centerline

K_STRAIGHT = 0.021     # |curvature| below this = straight-ish
MIN_STRAIGHT_M = 20    # straight-ish stretch must be at least this long
K_CORNER = 0.0385      # |curvature| above this = clearly a corner
SMOOTH_WINDOW_M = 5    # rolling-mean window (odd number, 1 point = 1 m)
K_LINK = 0.025         # curvature between two apexes must stay above this to be one corner
MAX_LINK_M = 35        # and the apexes must be at most this far apart



cl = load_centerline()
cl["curvature_smooth"] = smooth_circular(cl["curvature"].to_numpy(), SMOOTH_WINDOW_M)
curv = cl["curvature_smooth"].abs().to_numpy()      # |curvature|, one value per metre
n = len(curv)

# --- straights: low curvature for at least MIN_STRAIGHT_M ---
low = curv < K_STRAIGHT
low_runs = runs_of(low)
cl["is_straight"] = low & (np.bincount(low_runs)[low_runs] >= MIN_STRAIGHT_M)

# --- apexes: one per run of high curvature, at its sharpest point ---
high = curv > K_CORNER
cl["high_run"] = runs_of(high)
apex_idx = cl[high].groupby("high_run")["curvature_smooth"].apply(lambda c: c.abs().idxmax())
apexes = cl.loc[apex_idx].sort_values("s").reset_index(drop=True)
apexes["direction"] = np.where(apexes["curvature_smooth"] > 0, "L", "R")

# --- group apexes into corners ---
def same_corner(a, b) -> bool:
    """Apex b follows apex a: same direction, close enough, and the track keeps turning between them."""
    if a["direction"] != b["direction"]:
        return False
    i, j = int(a["s"]), int(b["s"])
    between = curv[i:j + 1] if j >= i else np.concatenate([curv[i:], curv[:j + 1]])  # wrap
    distance = (j - i) % n
    return distance <= MAX_LINK_M and between.min() >= K_LINK

corner_id = [0]
for k in range(1, len(apexes)):
    if same_corner(apexes.iloc[k - 1], apexes.iloc[k]):
        corner_id.append(corner_id[-1])
    else:
        corner_id.append(corner_id[-1] + 1)
corner_id = np.array(corner_id)

# loop: last apex and first apex may be the same corner across the start/finish line
if len(apexes) > 1 and same_corner(apexes.iloc[-1], apexes.iloc[0]):
    corner_id[corner_id == corner_id[-1]] = corner_id[0]

apexes["corner"] = pd.factorize(corner_id)[0] + 1      # renumber 1..N in order

# main apex of each corner = its sharpest apex
main_idx = apexes.groupby("corner")["curvature_smooth"].apply(lambda c: c.abs().idxmax())
apexes["is_main"] = apexes.index.isin(main_idx)

# --- plot ---
fig = go.Figure()
for is_straight, colour, name in [(True, "blue", "straight"), (False, "red", "corner")]:
    part = cl[cl["is_straight"] == is_straight]
    fig.add_trace(go.Scatter(
        x=part["x"], y=part["y"], mode="markers",
        marker=dict(size=4, color=colour),
        customdata=part[["s", "curvature_smooth"]],
        hovertemplate="s=%{customdata[0]:.0f} m<br>curvature=%{customdata[1]:.04f}<extra></extra>",
        name=name,
    ))

sharp = cl[high]
fig.add_trace(go.Scatter(
    x=sharp["x"], y=sharp["y"], mode="markers",
    marker=dict(size=3, color="black", symbol="x"),
    hoverinfo="skip", showlegend=False,
))

for is_main, size, colour, name in [(True, 11, "gold", "main apex"), (False, 7, "green", "other apex")]:
    part = apexes[apexes["is_main"] == is_main]
    fig.add_trace(go.Scatter(
        x=part["x"], y=part["y"],
        mode="markers+text" if is_main else "markers",
        marker=dict(size=size, color=colour, line=dict(width=1, color="black")),
        text=part["corner"].astype(str) + part["direction"],
        textposition="top center",
        customdata=part[["corner", "s", "curvature_smooth"]],
        hovertemplate="corner %{customdata[0]}<br>apex s=%{customdata[1]:.0f} m"
                      "<br>curvature=%{customdata[2]:.04f}<extra></extra>",
        name=name,
    ))

fig.update_yaxes(scaleanchor="x", scaleratio=1)
fig.update_layout(title=f"Corners (smoothed {SMOOTH_WINDOW_M} m, peak > {K_CORNER}, "
                        f"linked if curvature ≥ {K_LINK} and ≤ {MAX_LINK_M} m apart)",
                  xaxis_title="x (m, east)", yaxis_title="y (m, north)",
                  height=700, width=1000, legend=dict(orientation="h", y=-0.1))
fig.show()

print(f"{apexes['corner'].nunique()} corners, {len(apexes)} apexes")
```

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

from src.loader import load_centerline
from src.segment import find_corners, build_zones, label_zones
from src.config import PROC_DATA

THROTTLE = "pedal_throttle__throttle"
BRAKE = "pedal_brakes_pressure__front"
STEER = "steer_angle__angle"

grid = pd.read_parquet(PROC_DATA /"laps_distance.parquet")

cl = load_centerline()
cl, apexes = find_corners(cl)
cl, zones = build_zones(cl, apexes)
grid = label_zones(grid, cl, zones)
if "type" not in grid.columns:
    grid = grid.merge(zones[["zone", "type"]], on="zone", how="left")

# steering on straights only, as an absolute value
straights = grid[grid["type"] == "straight"].copy()
straights["abs_steer"] = straights[STEER].abs()

STINT_COLORS = {1: "#1f77b4", 2: "#ff7f0e"}


def overlay_hist(data, col, bin_width, title, xlabel, log_y=False):
    """One figure: one histogram per stint, overlaid, as % of that stint's points."""
    values = data[col].dropna()
    fig = go.Figure()
    for stint, g in data.groupby("stint"):
        fig.add_trace(
            go.Histogram(
                x=g[col].dropna(),
                xbins=dict(start=values.min(), end=values.max() + bin_width,
                           size=bin_width),
                histnorm="percent",
                opacity=0.5,
                name=f"stint {stint}",
                marker_color=STINT_COLORS.get(stint),
            )
        )
    fig.update_layout(
        title=title,
        xaxis_title=xlabel,
        yaxis_title="% of grid points in the stint",
        barmode="overlay",
        height=650,
        width=1600,
        font=dict(size=15),
    )
    if log_y:
        fig.update_yaxes(type="log")
    fig.show()


overlay_hist(grid, THROTTLE, bin_width=0.01,
             title="Throttle, whole lap", xlabel="throttle (0–1)", log_y=True)

overlay_hist(grid, BRAKE, bin_width=0.25,
             title="Front brake pressure, whole lap", xlabel="pressure (bar)", log_y=True)

overlay_hist(straights, "abs_steer", bin_width=0.25,
             title="|Steering|, straights only", xlabel="|steering angle| (deg)")
```

the rate change is next, because sensors jitter a lot, for example when a steering is held straigh the value tends to jump up and down by a value, like +/- 5 degrees. same thing for throttle and the brake pressure. so let's look at the rate change over two similar laps. let's take the lap in distance (m) not in time so we can see how the rate changes in unit/m

```python
GRID_STEP_M = 0.3
SMOOTH_M = 2.4 # smoothing window in metres
PAIR = (9, 19) # stint 1 lap, stint 2 lap
LAP_COLORS = {PAIR[0]: "#1f77b4", PAIR[1]: "#ff7f0e"}
LAP_DASH = {PAIR[0]: "solid", PAIR[1]: "dash"}

window = max(1, int(round(SMOOTH_M / GRID_STEP_M)))

RATE_SIGNALS = {
    "throttle": (THROTTLE, "throttle rate (1/m)"),
    "brake": (BRAKE, "brake rate (bar/m)"),
    "steer": ("abs_steer", "|steering| rate (deg/m)"),
}


def lap_rates(lap):
    """Raw and smoothed per-metre rates for one lap."""
    d = plot_df[plot_df["lap"] == lap].sort_values("s").copy()
    for key, (col, _) in RATE_SIGNALS.items():
        raw = d[col].to_numpy()
        smooth = (d[col].rolling(window, center=True, min_periods=1)
                  .mean().to_numpy())
        d[f"{key}_rate_raw"] = np.gradient(raw, d["s"].to_numpy())
        d[f"{key}_rate_smooth"] = np.gradient(smooth, d["s"].to_numpy())
    return d


rates = {lap: lap_rates(lap) for lap in PAIR}


def plot_rates_sector(sector, s_min, s_max):
    fig = make_subplots(
        rows=3, cols=1, shared_xaxes=True, vertical_spacing=0.04,
        subplot_titles=[label for _, label in RATE_SIGNALS.values()],
    )

    for lap in PAIR:
        d = rates[lap]
        d = d[(d["s"] >= s_min) & (d["s"] <= s_max)]
        stint = int(d["stint"].iloc[0])
        for row, key in enumerate(RATE_SIGNALS, start=1):
            fig.add_trace(
                go.Scatter(x=d["s"], y=d[f"{key}_rate_raw"], mode="lines",
                           name=f"lap {lap} (stint {stint}) raw",
                           line=dict(color=LAP_COLORS[lap], width=0.8,
                                     dash=LAP_DASH[lap]),
                           opacity=0.35,
                           legendgroup=f"{lap}raw", showlegend=(row == 1)),
                row=row, col=1,
            )
            fig.add_trace(
                go.Scatter(x=d["s"], y=d[f"{key}_rate_smooth"], mode="lines",
                           name=f"lap {lap} (stint {stint}) smoothed {SMOOTH_M:g} m",
                           line=dict(color=LAP_COLORS[lap], width=2.2,
                                     dash=LAP_DASH[lap]),
                           legendgroup=f"{lap}smooth", showlegend=(row == 1)),
                row=row, col=1,
            )

    for _, z in corner_zones.iterrows():
        if z["s_end"] < s_min or z["s_start"] > s_max:
            continue
        for row in (1, 2, 3):
            fig.add_vrect(x0=z["s_start"], x1=z["s_end"], fillcolor="grey",
                          opacity=0.12, line_width=0, row=row, col=1)
        fig.add_annotation(x=(z["s_start"] + z["s_end"]) / 2, y=1.02, yref="paper",
                           text=f"C{int(z['corner'])}", showarrow=False,
                           font=dict(size=12))

    for row in (1, 2, 3):
        fig.add_hline(y=0, line_dash="dot", line_color="black", row=row, col=1)

    fig.update_xaxes(title_text="s (m)", row=3, col=1)
    fig.update_layout(
        title=f"Rates, sector {sector} ({s_min:.0f}–{s_max:.0f} m), laps {PAIR[0]} vs {PAIR[1]}",
        height=950, width=1600, font=dict(size=14), hovermode="x unified",
    )
    fig.show()


for sector, (s_min, s_max) in SECTORS.items():
    plot_rates_sector(sector, s_min, s_max)
```

after experimenting with multiple values for the smoothing rate, i think the value of 2.4m is decet (so with the grid being 0.3m that is 8 grid points) is decent enough to flatten the areas with jitter to settle around 0 and at the same time keep the the trend where there is actual change (turning, braking, pushing on a straight) valid. 

next, i will investiate what could be considered a good value for the deadband, i.e. the rate change for each of these sensr values that will be considered negligible, sensor noice. for example, if the steering jumps up x degrees in an area where it should be steady, then we still consider the steering value to be the same across that period and neglect that change. hope that makes sense. 

for the steering, we will look at the steering behavior when the car is on a straight. for the braking and the throttle we will look at the data when the driver is applying them

```python
SMOOTH_M = 2.4
EXCLUDE_LAPS = [2] # standstills

LAT_ACC = "imu_acceleration__y"
SPEED = "vehicle_speed__u"


def add_rates(df, smooth_m=SMOOTH_M):
    """Per-lap smoothed per-metre rates for throttle, brake and |steering|."""
    w = max(1, int(round(smooth_m / GRID_STEP_M)))
    out = []
    for lap, d in df.groupby("lap"):
        d = d.sort_values("s").copy()
        s = d["s"].to_numpy()
        for col, key in [(THROTTLE, "thr"), (BRAKE, "brk"), ("abs_steer", "str")]:
            smooth = d[col].rolling(w, center=True, min_periods=1).mean()
            d[f"{key}_rate"] = np.gradient(smooth.to_numpy(), s)
        out.append(d)
    return pd.concat(out)


rated = add_rates(plot_df[~plot_df["lap"].isin(EXCLUDE_LAPS)])

# method 1: steering wobble when the car is really going straight 
held_straight = rated[
    (rated["type"] == "straight")
    & (rated[LAT_ACC].abs() < 0.03)
    & (rated[SPEED] > 8.0)
]
print("Method 1, |steering rate| (deg/m) when going straight")
print(
    held_straight.groupby("stint")["str_rate"]
    .apply(lambda r: r.abs().quantile([0.5, 0.9, 0.95, 0.99]))
    .unstack().round(2)
)
print("all:", held_straight["str_rate"].abs().quantile([0.5, 0.9, 0.95, 0.99]).round(2).to_dict())


# method 2: histogram of the rate where the signal is on 
def rate_hist(data, col, bin_width, x_max, title, xlabel):
    fig = go.Figure()
    for stint, g in data.groupby("stint"):
        fig.add_trace(go.Histogram(
            x=g[col].dropna(),
            xbins=dict(start=-x_max, end=x_max, size=bin_width),
            histnorm="percent", opacity=0.5, name=f"stint {stint}",
            marker_color=STINT_COLORS.get(stint),
        ))
    fig.add_vline(x=0, line_dash="dot", line_color="black")
    fig.update_layout(title=title, xaxis_title=xlabel,
                      yaxis_title="% of points in the stint",
                      barmode="overlay", height=650, width=1600, font=dict(size=14))
    fig.update_xaxes(range=[-x_max, x_max])
    fig.show()


rate_hist(rated[rated[THROTTLE] > THROTTLE_OFF], "thr_rate",
          bin_width=0.002, x_max=0.3,
          title=f"Throttle rate where throttle is on (smoothing {SMOOTH_M} m)",
          xlabel="throttle rate (1/m)")

rate_hist(rated[rated[BRAKE] > BRAKE_OFF], "brk_rate",
          bin_width=0.1, x_max=6,
          title=f"Brake rate where brake is on (smoothing {SMOOTH_M} m)",
          xlabel="brake rate (bar/m)")

rate_hist(rated, "str_rate",
          bin_width=0.5, x_max=40,
          title=f"|Steering| rate, all points (smoothing {SMOOTH_M} m)",
          xlabel="|steering| rate (deg/m)")
```

the first and second method agree for the steering deadband to be ~+/-2.5deg/m. as for the brake, from eyeballing it it looks like most jitter is about 1bar/m and for the throttle it looks like it's about 0.02/m.

let's try to find the width of the central peak and see if we mathematically backup the claims of methods 2. to do that we use the median absolute deviation (MAD)

```python
def mad_deadband(rate):
    """Robust width of the central peak of a rate distribution."""
    r = rate.dropna()
    mad = (r - r.median()).abs().median()
    spread = 1.4826 * mad
    return pd.Series({"spread": spread, "2x": 2 * spread, "3x": 3 * spread})


print("Method 3: MAD-based deadband on the 'on' points")
print(pd.DataFrame({
    "throttle (1/m)": mad_deadband(rated.loc[rated[THROTTLE] > THROTTLE_OFF, "thr_rate"]),
    "brake (bar/m)": mad_deadband(rated.loc[rated[BRAKE] > BRAKE_OFF, "brk_rate"]),
    "steering (deg/m)": mad_deadband(rated["str_rate"]),
}).round(3))

print("\nBrake noise floor: |brake rate| where brake is off (bar/m)")
print(rated.loc[rated[BRAKE] <= BRAKE_OFF, "brk_rate"].abs()
      .quantile([0.5, 0.9, 0.95, 0.99]).round(3).to_dict())
```

* throttle: the spread is 0.02 so it confirms what we saw in the plot.
* brake: the spread is 1.1, close to the 1bar/m we decided on, the noise in teh brake when the brake is off is around 0.3bar but that should not affect here because we took the brake phase to only include to be >0.5
* steering:  the value we chose (2.5) sit above the 2.2deg/m that we eyeballed but it's close enough.

the reason why i'm only paying attention to the 1x spread and not to the 1-3x is because the core of the rate distribution includes slow real movements, so the spread itself, checked against teh steering value since we know the deadband for it from method 1 and method 2, we can generalize over the pedals and assume that the spread is the deadband.

the values we finally settle for the deadband is:
* steering +/-2.5 deg/m 
* brake +/-1 bar/m 
* throttle +/-0.02 /m 

next, we need to decide on what's the minimum distance for a phase so that we have clean phases, we don't want to have 200 mini exists scattered around the track bc the conditions are met for .5 meters.

```python
GRID_STEP_M = 0.3
SMOOTH_M = 2.4

THROTTLE_OFF = 0.01
THROTTLE_FULL = 0.95
BRAKE_OFF = 0.5
STEER_ZERO_DEG = 10.0

# deadbands we agreed on
DB_THROTTLE = 0.02
DB_BRAKE = 1.0
DB_STEER = 2.5
EXCLUDE_LAPS = [2]
```

```python
def label_phases(df):
    """Assign one phase per grid point, first matching rule wins.
    Phases: ??, trail braking, braking, full throttle, coasting, lift, throttle on."""
    thr, brk, st = df[THROTTLE], df[BRAKE], df["abs_steer"]

    brake_on = brk > BRAKE_OFF
    thr_on = thr > THROTTLE_OFF
    full = thr >= THROTTLE_FULL
    steer_zero = st <= STEER_ZERO_DEG

    thr_rising = df["thr_rate"] > DB_THROTTLE
    thr_falling = df["thr_rate"] < -DB_THROTTLE
    brk_rising = df["brk_rate"] > DB_BRAKE
    brk_falling = df["brk_rate"] < -DB_BRAKE
    st_rising = df["str_rate"] > DB_STEER

    any_nan = df[[THROTTLE, BRAKE, "abs_steer",
                  "thr_rate", "brk_rate", "str_rate"]].isna().any(axis=1)

    conditions = [
        any_nan,
        brake_on & brk_rising & thr_on & thr_rising,             # ??
        brake_on & (~steer_zero | (brk_falling & st_rising)),    # trail braking
        brake_on,                                                # braking
        full,                                                    # full throttle
        ~thr_on & ~brake_on,                                     # coasting
        thr_falling & ~brake_on,                                 # lift
    ]
    choices = ["__nan__", "??", "trail braking", "braking", "full throttle",
               "coasting", "lift"]

    phase = np.select(conditions, choices, default="throttle on")
    df = df.copy()
    df["phase"] = pd.Series(phase, index=df.index).replace("__nan__", np.nan)
    return df
```

```python
def phase_runs(df):
    """Consecutive stretches of the same phase within each lap, with their length."""
    rows = []
    for lap, d in df.groupby("lap"):
        d = d.sort_values("s")
        run_id = (d["phase"] != d["phase"].shift()).cumsum()
        g = d.groupby(run_id).agg(phase=("phase", "first"),
                                  stint=("stint", "first"),
                                  s_start=("s", "first"),
                                  s_end=("s", "last"),
                                  n_points=("s", "size"))
        g["lap"] = lap
        rows.append(g)
    runs = pd.concat(rows, ignore_index=True).dropna(subset=["phase"])
    runs["length_m"] = runs["n_points"] * GRID_STEP_M
    return runs
```

```python
plot_df = grid.copy()
plot_df["abs_steer"] = plot_df[STEER].abs()
labelled = label_phases(add_rates(plot_df[~plot_df["lap"].isin(EXCLUDE_LAPS)]))
runs = phase_runs(labelled)

n_laps = labelled["lap"].nunique()
print(f"{len(runs)} runs over {n_laps} laps = one phase change every "
      f"{n_laps * 805 / len(runs):.1f} m on average\n")

summary = runs.groupby("phase").apply(lambda g: pd.Series({
    "runs": len(g),
    "median_m": g["length_m"].median(),
    "% runs < 1 m": (g["length_m"] < 1).mean() * 100,
    "% runs < 2 m": (g["length_m"] < 2).mean() * 100,
    "% distance in runs < 1 m": g.loc[g["length_m"] < 1, "length_m"].sum() / g["length_m"].sum() * 100,
    "% distance in runs < 2 m": g.loc[g["length_m"] < 2, "length_m"].sum() / g["length_m"].sum() * 100,
})).round(1)
print(summary.sort_values("runs", ascending=False).to_string())

fig = go.Figure()
for phase, g in runs.groupby("phase"):
    fig.add_trace(go.Histogram(x=g["length_m"], name=phase,
                               xbins=dict(start=0, end=20, size=GRID_STEP_M)))
for x in (1, 2):
    fig.add_vline(x=x, line_dash="dot", line_color="black")
fig.update_layout(title="Phase run lengths, all laps, final parameters",
                  xaxis_title="run length (m)", yaxis_title="number of runs",
                  barmode="stack", height=650, width=1600, font=dict(size=14))
fig.update_xaxes(range=[0, 20])
fig.show()
```
